```python
# 1. Read Parquet
df = spark.read.parquet(
    '/Volumes/dlt_sk/test_files/test_files/part-r-00000-1a9822ba-b8fb-4d8e-844a-ea30d0801b9e.gz.parquet'
)

# 2. Filter rows
df_filtered = df.filter(df["count"] > 100)

# 3. Select required columns
df_selected = df_filtered.select(
    "DEST_COUNTRY_NAME",
    "ORIGIN_COUNTRY_NAME",
    "count"
)

# 4. Remove duplicate records
df_unique = df_selected.dropDuplicates()

# 5. Group by destination country
df_grouped = df_unique.groupBy(
    "DEST_COUNTRY_NAME"
).sum("count")

# 6. Rename aggregated column
df_final = df_grouped.withColumnRenamed(
    "sum(count)",
    "total_flights"
)

# 7. Sort the result
df_final = df_final.orderBy(
    df_final["total_flights"].desc()
)

# 8. Trigger Spark job
df_final.show()

# 9. Write result
df_final.write \
    .format("csv") \
    .option("header", "true") \
    .mode("overwrite") \
    .save('/Volumes/dlt_sk/test_files/test_files/students_output')
```
