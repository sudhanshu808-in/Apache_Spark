```python
from pyspark.sql.functions import * ;
from pyspark.sql.types import * ;
```

```python
%sql
select * from dlt_sk.test_files.students;
```

```python
df =  spark.sql(""" select * from dlt_sk.test_files.students""")
df.show()
```

```python
df.withColumn("age",when(col("age")<=20,lit(1000)).otherwise(col("age")))\
    .withColumn("is_valid",when(col("age")<=20,lit("NO")).otherwise("yes")).show()
```

```python
df.withColumn("type",when((col("age")>=20) & (col("age")<=22) ,"Z")
              .when((col("student_id")>=5) &(col("student_id")<=10),"Pluhh..." )
              .otherwise("Unc"))\
              .show()
```
